### Implement Backpropagation Algorithm from Scratch
We will implement a simple Multi-Layer Perceptron (MLP) with a custom hidden layer configuration, using the sigmoid activation function, and train it using the Backpropagation algorithm.

```python
import numpy as np

class NeuralNetwork:
    def __init__(self, input_size, hidden_size, output_size, lr=0.1):
        self.lr = lr
        # Initialize weights randomly with standard normal distribution
        self.W1 = np.random.uniform(-1.0, 1.0, (input_size, hidden_size))
        self.b1 = np.zeros((1, hidden_size))
        self.W2 = np.random.uniform(-1.0, 1.0, (hidden_size, output_size))
        self.b2 = np.zeros((1, output_size))

    def sigmoid(self, x):
        return 1 / (1 + np.exp(-x))

    def sigmoid_derivative(self, x):
        # x is assumed to be the activation output: sigmoid(z)
        return x * (1 - x)

    def forward(self, X):
        self.z1 = np.dot(X, self.W1) + self.b1
        self.a1 = self.sigmoid(self.z1)
        self.z2 = np.dot(self.a1, self.W2) + self.b2
        self.a2 = self.sigmoid(self.z2)
        return self.a2

    def backward(self, X, y, output):
        # Loss error and derivative
        error = y - output
        d_output = error * self.sigmoid_derivative(output)

        # Error and delta for hidden layer
        error_hidden = np.dot(d_output, self.W2.T)
        d_hidden = error_hidden * self.sigmoid_derivative(self.a1)

        # Update weights and biases using gradient descent
        self.W2 += np.dot(self.a1.T, d_output) * self.lr
        self.b2 += np.sum(d_output, axis=0, keepdims=True) * self.lr
        self.W1 += np.dot(X.T, d_hidden) * self.lr
        self.b1 += np.sum(d_hidden, axis=0, keepdims=True) * self.lr

    def train(self, X, y, epochs=10000):
        for epoch in range(epochs):
            output = self.forward(X)
            self.backward(X, y, output)
            if epoch % (epochs // 10) == 0:
                loss = np.mean(np.square(y - output))
                print(f"Epoch {epoch}/{epochs} - Loss: {loss:.6f}")
```

### Define a Toy Dataset and Test the Network
Now we will define the input data representing the XOR function, train the model, and verify the resulting predictions.

```python
# XOR dataset
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
y = np.array([[0], [1], [1], [0]])

# Initialize and train the neural network
nn = NeuralNetwork(input_size=2, hidden_size=4, output_size=1, lr=0.2)
nn.train(X, y, epochs=20000)

# Test the trained model
print("\nPredictions after training:")
for idx in range(len(X)):
    pred = nn.forward(X[idx:idx+1])
    predicted_class = int(pred[0][0] > 0.5)
    print(f"Input: {X[idx]} -> Target: {y[idx][0]} -> Prediction: {pred[0][0]:.4f} (Class: {predicted_class})")
```
