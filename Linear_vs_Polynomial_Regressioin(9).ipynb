```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures

X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)
y = np.array([2, 4, 5, 4, 8])

linear = LinearRegression()
linear.fit(X, y)
y_linear = linear.predict(X)

poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(X)

model = LinearRegression()
model.fit(X_poly, y)
y_poly = model.predict(X_poly)

plt.scatter(X, y, color='blue', label='Actual Data')
plt.plot(X, y_linear, color='red', label='Linear Regression')
plt.plot(X, y_poly, color='green', label='Polynomial Regression')

plt.xlabel('X Values')
plt.ylabel('Y Values')
plt.title('Linear vs Polynomial Regression')
plt.legend()
plt.show()

print("Linear Predictions:", y_linear)
print("Polynomial Predictions:", y_poly)
```
