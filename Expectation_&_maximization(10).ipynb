```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.mixture import GaussianMixture

X = np.array([[1], [2], [3], [8], [9], [10]])

model = GaussianMixture(n_components=2, random_state=0)
model.fit(X)

labels = model.predict(X)

print("Cluster Labels:", labels)
print("Mean:", model.means_)
print("Covariance:", model.covariances_)

plt.scatter(X, np.zeros(len(X)), c=labels)
plt.title("Expectation Maximization Algorithm")
plt.xlabel("Data Values")
plt.show()
```
