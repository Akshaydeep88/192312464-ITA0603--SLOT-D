```python

import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

X = np.array([10, 20, 30, 40, 50, 60]).reshape(-1, 1)
y = np.array([9, 8, 7, 6, 5, 4])

model = LinearRegression()
model.fit(X, y)

pred = model.predict([[35]])

plt.scatter(X, y, label="Actual Prices")
plt.plot(X, model.predict(X), label="Regression Line")
plt.scatter(35, pred, color="red", label="Predicted Price")

plt.title("Car Price Prediction")
plt.xlabel("Mileage (km/l)")
plt.ylabel("Price (Lakhs)")
plt.legend()
plt.show()

print("Predicted Price:", pred[0], "Lakhs")
```
