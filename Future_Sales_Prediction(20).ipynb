```python

import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

X = np.array([1, 2, 3, 4, 5, 6, 7, 8]).reshape(-1, 1)

y = np.array([100, 120, 130, 150, 170, 180, 200, 220])

model = LinearRegression()
model.fit(X, y)

future = np.arange(1, 13).reshape(-1, 1)
pred = model.predict(future)

plt.scatter(X, y, label="Actual Sales")
plt.plot(future, pred, "r--", label="Predicted Sales")

plt.title("Future Sales Prediction using Linear Regression")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.legend()
plt.show()

print("Future Sales Prediction:")

for i in range(8, 12):
    print("Month", i+1, ":", round(pred[i], 2))
```
