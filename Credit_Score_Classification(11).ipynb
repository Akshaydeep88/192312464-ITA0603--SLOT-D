```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn.preprocessing import LabelEncoder

data = {
    'Income': [25000, 40000, 60000, 80000, 30000, 90000],
    'Debt': [20000, 15000, 10000, 5000, 25000, 3000],
    'Payment_History': ['Poor', 'Average', 'Good', 'Good', 'Poor', 'Good'],
    'Credit_Score': ['Low', 'Medium', 'High', 'High', 'Low', 'High']
}

df = pd.DataFrame(data)

le = LabelEncoder()
df['Payment_History'] = le.fit_transform(df['Payment_History'])

X = df[['Income', 'Debt', 'Payment_History']]
y = df['Credit_Score']

model = DecisionTreeClassifier(random_state=0)
model.fit(X, y)

prediction = model.predict([[50000, 12000, le.transform(['Good'])[0]]])

print("Predicted Credit Score:", prediction[0])
```
