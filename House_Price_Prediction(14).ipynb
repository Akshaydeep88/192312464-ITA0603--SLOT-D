```python

import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

X = np.array([800, 1000, 1200, 1400, 1600, 1800, 2000]).reshape(-1, 1)
y = np.array([20, 25, 30, 38, 40, 45, 50])

model = LinearRegression()
model.fit(X, y)

pred = model.predict([[1600]])

plt.scatter(X, y, label="Actual Prices")
plt.plot(X, model.predict(X), label="Regression Line")
plt.scatter(1600, pred, color="red", label="Predicted Price")

plt.title("House Price Prediction")
plt.xlabel("Area (sq.ft)")
plt.ylabel("Price (Lakhs)")
plt.legend()
plt.show()

print("Predicted Price:", pred[0], "Lakhs")
```
